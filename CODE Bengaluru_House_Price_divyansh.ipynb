```python
import pandas as pd

df = pd.read_csv("/content/bengaluru_house_prices.csv")

df.head()
```

```python
df = df.drop(['area_type'], axis=1)
df.head()
```

```python
df = df.drop(['availability'], axis=1)
df.head()
```

```python
df=df.drop(['society'],axis=1)
df.head()
```

```python
df=df.drop(['balcony'],axis=1)
df.head()
```

```python
df['bhk'] = df['size'].astype(str).apply(lambda x: int(x.split()[0]) if isinstance(x, str) and x.split()[0].isdigit() else None)

df = df.dropna(subset=['bhk'])

df.head()
```

```python
df=df.drop(['size'],axis=1)
df.head()
```

```python
def convert_sqft(x):
    try:
        if '-' in str(x):
            a, b = x.split('-')
            return (float(a) + float(b)) / 2
        return float(x)
    except:
        return None

df['total_sqft'] = df['total_sqft'].apply(convert_sqft)

df = df.dropna(subset=['total_sqft'])

df.head()
```

```python
clean_path = "/content/cleaned_bengaluru.csv"
df.to_csv(clean_path, index=False)

print("✅ Cleaned dataset saved successfully at:", clean_path)

df.head()
```

```python

import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error
import joblib
import warnings
warnings.filterwarnings("ignore")

DATA_PATH = "/content/cleaned_bengaluru.csv"
if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(f"Cleaned file not found at {DATA_PATH}. Upload it to Colab or change the path.")

df = pd.read_csv(DATA_PATH)
print("Loaded dataset rows:", len(df))
print(df.head())

FEATURES = ['location', 'bhk', 'total_sqft', 'bath']
TARGET = 'price'

df = df.dropna(subset=FEATURES + [TARGET]).reset_index(drop=True)

X = df[FEATURES].copy()
y = df[TARGET].copy()

X['bhk'] = pd.to_numeric(X['bhk'], errors='coerce')
X['total_sqft'] = pd.to_numeric(X['total_sqft'], errors='coerce')
X['bath'] = pd.to_numeric(X['bath'], errors='coerce')

good = X.notnull().all(axis=1)
X = X[good].reset_index(drop=True)
y = y[good].reset_index(drop=True)

print("\nFeature stats:")
print(X.describe())

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print("\nTrain rows:", len(X_train), "Test rows:", len(X_test))

categorical_features = ['location']
numeric_features = ['bhk', 'total_sqft', 'bath']

try:
    enc = OneHotEncoder(handle_unknown='ignore', sparse=False)
except TypeError:
    enc = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

preprocessor = ColumnTransformer(transformers=[
    ('cat', enc, categorical_features),
    ('num', StandardScaler(), numeric_features)
])

model = Pipeline(steps=[
    ('pre', preprocessor),
    ('reg', LinearRegression())
])

print("\nTraining model...")
model.fit(X_train, y_train)
print("Model trained.")

y_pred = model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"\nTest RMSE: {rmse:.4f}  (units same as 'price' column)")

comp = pd.DataFrame({
    'location': X_test['location'].values,
    'bhk': X_test['bhk'].values,
    'total_sqft': X_test['total_sqft'].values,
    'bath': X_test['bath'].values,
    'actual_price': y_test.values,
    'pred_price': np.round(y_pred, 3)
})
print("\nSample predictions:")
print(comp.head(8))

OUT_MODEL = "/content/bengaluru_price_model.joblib"
joblib.dump(model, OUT_MODEL)
print(f"\nSaved pipeline to: {OUT_MODEL}")

def ask_and_predict(pipeline):
    """Interactive loop to input features and get predicted price."""
    print("\n--- Interactive prediction ---")
    print("Type 'exit' to stop.")
    while True:
        s = input("Enter BHK (e.g. 2) or 'exit': ").strip()
        if s.lower() == 'exit':
            break
        try:
            bhk = float(s)
        except:
            print("Invalid bhk. Try again.")
            continue

        loc = input("Enter location (exact or new name): ").strip()
        sqft_s = input("Enter total_sqft (e.g. 1200): ").strip()
        bath_s = input("Enter bath (e.g. 2): ").strip()
        try:
            sqft = float(sqft_s)
            bath = float(bath_s)
        except:
            print("Invalid numeric input for sqft or bath. Try again.")
            continue

        sample = pd.DataFrame([{
            'location': loc,
            'bhk': bhk,
            'total_sqft': sqft,
            'bath': bath
        }])
        pred = pipeline.predict(sample)[0]
        print(f"Predicted price: {pred:.3f} (same units as dataset)\n")

print("\nTo try interactive prediction, call: ask_and_predict(model)")
```
